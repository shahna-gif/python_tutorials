#### 27. Given: 
inventory = { 
"Laptop": 3,

"Keyboard": 4, 
"Monitor": 0 
} 
Ask the user for a product. 
Rules: 
0 → "Out of Stock" 
1–3 → "Reorder" 
4–10 → "Monitor Stock" 
Above 10 → "Stock Sufficient"
If the product has 1–3 units, the desired stock level is 10 units. Calculate how many units need to be ordered.


```python
inventory = { "Laptop": 3, "Mouse": 25, "Keyboard": 4, "Monitor": 0 } 
product = input("enter a product").title()
if product in inventory:
    stock= inventory[product]
    if stock==0:
        print("out of stock")
    elif stock>=1 and stock<=3:
        print("reorder")

        order = 10 - stock
        print("units to order:",order)
    elif stock >=4 and stock<=10:
        print("monitor stock")
    else:
        print("stock sufficient")
else:
    print("product not found")
```

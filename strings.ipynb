## String Practical Problems

#### Q1- Create a string and print its first and last character

```python
Text = "Shahna"
print(Text)
print("first letter:",Text[0])
print("last letter:",Text[-1])
```

#### Q2- Count the number of occurrences of "a" in "banana".

```python
Text = "banana"
print(Text)
print(Text.count("a"))
print(Text.count("h"))
```

#### Q3- Reverse a string using slicer

```python
Text = "Python"
print(Text)
print("Reverse of the text:",Text[ : :-1])
```

#### Q4- Replace "Java" with "Python" from "Java is simple & powerful".

```python
Text = "Java is simple & Powerful"
print(Text)
print(Text.replace("Java","Python"))
```

#### Q5- convert a string to uppercase and lowercase

```python
Text = "Shahna k"
print(Text)
print(Text.upper())
print(Text.lower())
print(Text.capitalize())
print(Text.title())
```

#### Q6-Split "Python,Java,SQL" into a list

```python
Text = "Python,Java,SQL"
print(Text)
print(Text.split(","))
```

#### Q7-Join ["Python","Java","SQL"] using "-"

```python
Text = ["Python","Java","SQL"]
print(Text)
print("-".join(Text))
```

#### Q8- check whether "thon" exist in"python"

```python
Text = "Python"
print(Text)
print("thon"in Text)
```

#### Q9- print every second character from "Programming"


```python
Text = "programming"
print(Text)
print("Every second character:",Text[ : :2])
```

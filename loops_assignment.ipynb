##### 14. Take a number and check whether it is an **Armstrong number**.
    Example: `153 → Armstrong number`


```python
num = int(input("enter the number:"))
num = str(num)
digits = len(num)
sum = 0
for i in num:
    power =int(i)**digits
    #print("power of digits:",power)
    sum = sum + power
    #print("sum of the numbers:",sum)
if sum == int(num):
    print("it's an armstrong number")
else:
    print("it's not an armstrong number ")    
```

##### 18. Take two numbers from the user and find their **GCD/HCF** using a loop.


```python
num1 = int(input("enter the number:"))
num2 = int(input("enter the number:"))
gcd = 1
for i in range(1, min(num1,num2)+1):
    if num1%i==0 and num2%i==0:
        gcd = i
print("GCD/HCF:",gcd)
```

##### 27. Create a simple **ATM PIN checker**.
    Give the user a maximum of **3 attempts** to enter the correct PIN. Display `"Account Locked"` after 3 incorrect attempts.


```python
correct_pin = 1234
attempts = 0
while attempts < 3:
    pin = int(input("enter the pin:"))

    if pin == correct_pin:
        print("correct pin")
        break
    else:
        attempts = attempts+1
        print("incorrect pin")
if attempts == 3:
    print("Account blocked")   


```

##### 28. Create a **login system** where the user gets **3 attempts** to enter the correct username and password.


```python
correct_username = "admin"
correct_password = "1234"
attempts = 0
while attempts < 3:
    username = input("enter username:")
    password = input("enter passsword:")
    if username == correct_username and password == correct_password:
        print("login succesful")
        break
    else:
        attempts = attempts + 1
        print("invalid username or password")    

if attempts == 3:
    print("account locked")        

```

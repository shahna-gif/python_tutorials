##### loop with datastructures

```python
#find double of each number
number = [23,45,23,1,4,87,5]
for num in number:
    print(num*2)
```

```python
# create a new list
number = [23,45,23,1,4,87,5]
double = []
for num in number:
    double.append(num*2)
print(double)
    
    
```

```python
#create a new list of numbers which is multiple of 3 from 30 to 60
list = []
for i in range(30,60):
    if i%3==0:
        list.append(i)
print(list)        
```

```python
#create a list of first n natural numbers
number = int(input("enter the first n naltural numbers:"))
list = []
for i in range(1,number+1):
    list.append(i)
print(list)    
```

```python
#create a list with length of each element in given list
fruits = ["apple","banana","mango","grapes"]
list = []
for fruit in fruits:
    list.append(len(fruit))
print(list)
```

```python
d = {}
for fruit in fruits:
  d[fruit]=len(fruit)
print(d)
```

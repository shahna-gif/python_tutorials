#### 1. Store your date of birth as a tuple in the format (day, month, year) and print it. 


```python
Date_of_birth = (14,1,2004)
print(Date_of_birth)
```

#### 2. A GPS location is given a (12.9716, 77.5946).Print the latitude and the longitude separately. 


```python
Location = (12.9716, 77.5946)
print(Location[0])
print(Location[1])
```

#### 3. A product's details are stored as ("Laptop", 55000, "Electronics"). Print the price alone. 


```python
Product_details = ("Laptop", 55000, "Electronics")
print(Product_details[1]) 
```

#### 4. Try changing the price in the tuple from the previous question directly to 60000. Run it, see what happens, and write down the result as a comment in your code. 


```python
Product_details = ("Laptop", 55000, "Electronics")
Product_details[1] = 60000
print(Product_details)
# Result:tuple values are immutable
```

#### 5. A traffic light shows only three states: ("Red", "Yellow", "Green"). Print how many states it contains. 


```python
Traffic_light = ("Red","Yellow","Green")
print(len(Traffic_light))
```

#### 6. A movie's rating history is (4, 5, 3, 5, 5, 2). Print how many times it was rated 5

```python
Movies_rating = (4, 5, 3, 5, 5, 2)
print(Movies_rating.count(5))
```

#### 9. A store's opening and closing time are stored together as ("09:00 AM", "09:00 PM"). Print the position of "09:00 PM" in the tuple.

#### 10. Two friends share their favourite colours as color1 = ("Blue", "Green") and color2 = ("Red", "Yellow"). Combine both into a single tuple and print it. 


```python
color1 = ("Blue", "Green")
color2 = ("Red", "Yellow")
color = color1 + color2
print(color)
```

#### 11. A recipe needs exactly three ingredients in a fixed order: ("Flour", "Sugar", "Eggs"). Print the ingredients in reverse order without changing the original tuple. 


```python
ingredients = ("Flour", "Sugar", "Eggs")
print(ingredients)
print(ingredients[ : :-1])
```

#### 12. A student's marks in 3 subjects are given as (78, 85, 92). Print the total and the average.


```python
Marks = (78,85,92)
total = sum(Marks)
average = total/len(Marks)
print("Total:",total)
print("Average score:",average)
```

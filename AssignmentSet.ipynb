#### 1. A survey asked 6 people their favourite programming language: ["Python", "Java", "Python", "C++", "Java", "Python"]. Store the responses and print only the distinct languages mentioned. 


```python
Language = ["Python", "Java", "Python", "C++", "Java", "Python"]
print(Language)
print(set(Language))

```

#### 2. Two gyms report their members' names: gymA = {"Ravi", "Sneha", "Kiran", "Alok"} and gymB = {"Kiran", "Meera", "Alok", "Priya"}. Print everyone who is a member of at least one gym. 


```python
gymA = {"Ravi", "Sneha", "Kiran", "Alok"}
gymB = {"Kiran", "Meera", "Alok", "Priya"}
all_members = gymA | gymB
print(all_members)
```

#### 4. Using the same two gyms above, print the names of people who are members of gymA only.

#### 6. A trip itinerary visited these cities in order: ("Delhi", "Agra", "Jaipur", "Delhi", "Agra"). Print how many distinct cities were actually visited. 


```python
cities = ("Delhi", "Agra", "Jaipur", "Delhi", "Agra")
distint_cities = set(cities)
print(len(distint_cities))
```

#### 7. A teacher wants to combine two attendance sheets recorded separately as lists: morning = ["Ana", "Ben", "Cara", "Ana"] and afternoon = ["Ben", "Dev", "Cara"]. Print the names of students who attended only in the morning. 


```python
morning = ["Ana", "Ben", "Cara", "Ana"]
afternoon = ["Ben", "Dev", "Cara"]
print(set(morning)-set(afternoon))
```

# Conditional Statements in Python — Practical Notebook

Conditional statements let your program make decisions — running different code depending on whether something is `True` or `False`. This is the first building block you'll use constantly in data work: checking values, flagging missing data, validating inputs, and classifying results.

## `if` — run code only when a condition is true

If the condition is `True`, the indented block runs. If it's `False`, Python simply skips it.

```python
age = int(input("Enter your age: "))
if age >= 18:
    print("You are eligible to vote")
```

## `if` / `else` — choose between two outcomes

When the condition is `True`, the `if` block runs. When it's `False`, the `else` block runs instead. Exactly one of the two always executes.

```python
marks = int(input("enter score: "))
if marks >= 40:
    print("Pass")
else:
    print("Fail")
```

## `if` / `elif` / `else` — choose between several outcomes

Python checks each condition in order and runs the first one that's `True`. If none of them are true, the final `else` runs.

```python
score = int(input("enter ur score: "))
if score >= 90:
    print("Grade A")
elif score >= 75:
    print("Grade B")
elif score >= 60:
    print("Grade C")
else:
    print("Grade D")
```

## Conditions inside conditions

A conditional block can contain another conditional inside it — useful when a decision depends on more than one check happening in sequence.

```python
age = 12
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("Entry denied")
```

## Combining more than one condition

You can join conditions together instead of nesting them, using the logical operators covered earlier.

```python
temperature = 30
is_raining = True

if temperature > 25 and not is_raining:
    print("Good day for a walk")
```

```python
cgpa = 8.2
backlogs = 0

if cgpa >= 7.5 or backlogs == 0:
    print("Eligible for placement")
```

## A data-checking example

Before doing anything with a value, it often helps to check what state it's in first — this is a pattern you'll use constantly when working with real datasets.

```python
column_value = None

if column_value is None:
    print("Missing value detected")
```

```python
current_type = "int64"
allowed_types = ["int64", "float64"]

if current_type in allowed_types:
    print("Valid numeric column type")
else:
    print("Unsupported column type")
```

## 1. Why do we need loops?

Imagine printing "Hello" 5 times. Without loops:
```python
print("Hello")
print("Hello")
print("Hello")
print("Hello")
print("Hello")
```
This is repetitive. **Loops** let us repeat a block of code without rewriting it.

Python has two main loops:
- **`for` loop** → used when we know how many times to repeat, or when we want to go through items one by one.
- **`while` loop** → used when we repeat *until a condition becomes False* (number of repetitions may not be known in advance).

````python
# Without loop (repetitive)
print("Hello")
print("Hello")
print("Hello")
````

````python
range(3)
````

````python
# With loop (clean)
for i in range(10):
    print("Hello")
````

## 2. `for` loop basics — using `range()`

`range(start, stop, step)` generates numbers. `stop` is **not included**.

- `range(5)` → 0,1,2,3,4
- `range(2,6)` → 2,3,4,5
- `range(0,10,2)` → 0,2,4,6,8

````python
# basic for loop with range
for i in range(5):
    print("Count:", i)
````

````python
# for loop with start, stop
for i in range(2, 6):
    print(i)
````

````python
# for loop with step (using arithmetic operator too)
for i in range(0, 10):
  print(i, "-> square is", i * i)   # using ** or * operator
````

````python
# for loop + conditional statement -> print even/odd
for i in range(1, 11):
    if i % 2 == 0:          # modulus operator
        print(i, "is Even")
    else:
        print(i, "is Odd")
````

````python
# multiplication table of 5
for i in range(1,11):
  print(i,"x 5 =",i*5)
````

````python
for i in range(10):
  print(i,end=" ")
````

## 4. `while` loop basics

`while condition:` keeps running the block **as long as the condition is True**.

⚠️ Make sure the condition eventually becomes False, otherwise it becomes an **infinite loop**.

````python
# basic while loop - countdown
n = 10
while n > 0:
    print(n)
    n -= 1          # important: update the variable, else infinite loop
print("Liftoff!")
````

````python
# while loop with input - keep asking until user enters 'stop'
# (Run this cell and type a few words, then type 'stop')
word = ""
while word.lower() != "stop":
    word = input("Enter a word (type 'stop' to end): ")
````

## 5. `break` and `continue`

- **`break`** → stops the loop immediately.
- **`continue`** → skips the current iteration and moves to the next one.

````python
# break example - stop loop when we find the number
numbers = [4, 8, 15, 16, 23, 42]
search = 16

for n in numbers:
    if n == search:
        print("Found", search, "in the list!")
        break
    print("Checked", n, "- not a match")
````

````python
# continue example - skip printing negative numbers
numbers = [10, -5, 20, -3, 8]

for n in numbers:
    if n < 0:
        continue        # skip rest of code for this value
    print(n)
````

## 6. Nested loops (loop inside a loop)

Useful for grids, tables, and patterns.

````python
# Outer loop runs 3 times
for i in range(3):

    # Inner loop runs 4 times for every outer loop iteration
    for j in range(4):
        print(j)
````

````python
# Outer loop controls the number of rows
for i in range(3):
    for j in range(4):
        print("*",end=" ")
    print( )    
````

````python
# multiplication table using nested for loops
for i in range(1,11):
    for j in range(1,11):
        print(i, "x", j, "=",i*j)
    print("----------")   
````

````python
# Number pyramid — each row prints DIFFERENT numbers
for i in range(1,5):
    for j in range(1,i+1):
        print(j, end=" ")
    print()    
````

````python
#  Hollow square — border is '*', inside is blank (needs per-position decision)
num = int(input ("enter a number:"))
for i in range(1,num+1):
    for j in range(1,num+1):
     if i==1 or i==num:
        print("*",end=" ")
     else:   
        if j==1 or j==num: 
          print("*",end=" ")
        else:
          print(" ",end=" ") 
    print( )       

                        
````

## 7. Practical Mini-Programs

Combining loops with input, conditionals, operators and data structures.

````python
# Practical 1: Simple Calculator that keeps running until the user chooses to exit 
while True:
    print("\n1-Addition,2-Substraction,3-Multiplication,4-Division,5-Exit")
    choice=int(input("choose an operation(1-5):"))
    if choice==5:
            print("Exit")
            break
    num1 = int(input("Enter the number:"))
    num2 = int(input("Enter the number:"))
    
    if choice==1:
        print("sum",num1+num2)    
    elif choice==2:
        print("Difference",num1-num2)
    elif choice==3:
        print("multiplictaion",num1*num2)
    else:
        if num2==0:
             print("error: division by zero") 
        else:
             print("division",num1/num2)            
    
````

````python
# Practical 2: Check if a number is prime (loops + conditionals + operators)
num = int(input("enter a number:"))
prime = True
for i in range(2,num):
  if num%i==0:
   prime = False
   print(" not prime number")
   break
if prime:
   print(num,"  prime number") 
else:
  print(num, "not  prime")      
````

````python
# Practical 3: Grade report card using dictionary + for loop + conditionals
Students = {"Anu":92,"Ravi":67,"Meeraa":45,"Kiran":78}
#98-A,70-B,30-C Other fall
for name,mark in Students.items():
    if mark>=90:
        print(name,"-Grade A-",mark)
    elif mark>= 70:
        print(name,"-Grade B-",mark)
    elif mark>=30:
        print(name,"-Grade C-",mark)
    else:
        print(name,"-Fail-",mark)           
````

````python
# Practical 4: Reverse a list manually using a while loop (without using [::-1])
value = []
prices = [30,46,2,9,7]
for i in range(len(prices)-1,-1,-1):
    value.append(prices[i])
print(value)
````

````python
# using while loop
value = []
prices = [30,46,2,9,7]
count = len(prices)
while count>0:
    count=count-1
    value.append(prices[count])
print(value)
````

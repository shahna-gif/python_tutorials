#### 2. A small shop's daily sales for a week are recorded as [1200, 1500, 900, 1700,

#### 3. A grocery list is ["Milk", "Bread", "Eggs"]. The customer forgot to add "Butter" —

#### 4. A teacher recorded these marks: [56, 78, 45, 90, 67]. Print the total and the average 
mark.  

```python
Marks = [56, 78, 45, 90, 67]
print("Total score:",sum(Marks))
print("Average score:",sum(Marks)/len(Marks))
```
